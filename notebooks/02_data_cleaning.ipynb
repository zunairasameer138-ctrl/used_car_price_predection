# 02 - Data Cleaning

Applies `src/data_cleaning.clean_data()` to the raw dataset and inspects exactly what it changed. See the docstring in that module for the reasoning behind every step.

```python
import sys
sys.path.insert(0, '../src')
import pandas as pd
from data_loader import load_raw_data
from data_cleaning import clean_data

raw = load_raw_data('../data/raw/CAR DETAILS FROM CAR DEKHO.csv')
clean, report = clean_data(raw)
print(f'{report.rows_in} raw rows -> {report.rows_out} clean rows')
```

## Full cleaning report

```python
for k, v in report.as_dict().items():
    print(f'{k}: {v}')
```

## Verify the `OpelCorsa` glitch was fixed

```python
assert clean['name'].str.startswith('OpelCorsa').sum() == 0
clean[clean['name'].str.startswith('Opel')][['name', 'year', 'selling_price']]
```

## Verify no duplicates remain

```python
assert clean.duplicated().sum() == 0
print('Duplicates remaining:', clean.duplicated().sum())
```
